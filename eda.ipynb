```python
import pandas as pd
```

```python
# Check for shapes of the datasets
books_tags_df = pd.read_csv('book_tags.csv')
print("Books tag shape:", books_tags_df.shape)
books_df = pd.read_csv('books.csv')
print("Books shape:", books_df.shape)
ratings_df = pd.read_csv('ratings.csv')
print("Rating shape:", ratings_df.shape)
tags_df = pd.read_csv('tags.csv')
print("Tag shape:", tags_df.shape)
to_read_df = pd.read_csv('to_read.csv')
print("To read shape:", to_read_df.shape)
```

```python
print("Books_tag columns:", books_tags_df.columns.to_list())
print("Books columns:", books_df.columns.to_list())
print("Rating columns:", ratings_df.columns.to_list())
print("Tag columns:", tags_df.columns.to_list())
print("To_read columns:", to_read_df.columns.to_list())
```

```python
# Merge ratings 
merged_df = pd.merge(ratings_df, to_read_df, on=['user_id', 'book_id'], how='outer')

# Merge with books on 'book_id
merged_df = pd.merge(merged_df, books_df, on='book_id', how='left')

# Merge with book_tags
merged_df = pd.merge(merged_df, books_tags_df, left_on='book_id', right_on='goodreads_book_id', how='left')

# Merge with tags 
merged_df = pd.merge(merged_df, tags_df, on='tag_id', how='left')

# Drop duplicate 
merged_df.drop(columns=['goodreads_book_id'], inplace=True, errors='ignore')

# Preview result
print("Merged dataset shape:", merged_df.shape)
print(merged_df.head())
```

```python
merged_df.dropna(inplace=True)
# Preview result
print("Merged dataset shape:", merged_df.shape)
print(merged_df.head())
```
